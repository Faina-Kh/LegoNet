# Run LegoNet Main

This notebook is a small, safe wrapper around `legonet/scripts/main.py`. Edit the settings cell, preview the command, then run it.

```python
from pathlib import Path
import subprocess
import sys


def find_project_root(start: Path) -> Path:
    """Find the repository root from the current notebook working directory."""
    for path in (start, *start.parents):
        if (path / "legonet" / "scripts" / "main.py").exists():
            return path
    raise FileNotFoundError("Could not find legonet/scripts/main.py above the current directory.")


PROJECT_ROOT = find_project_root(Path.cwd())
MAIN_SCRIPT = PROJECT_ROOT / "legonet" / "scripts" / "main.py"

PROJECT_ROOT
```

## Settings

```python
# Update these values for your machine and experiment.
storage_path = r"C:\\Users\\bordezki\\Desktop\\LegoNet"
dataset_name = "roots"  # "roots" or "grapes"
network_type = "both_for_roots_2" # "counting_lean"  #"counting_reg" #"both_Back2bFind2b" #"both_for_roots_2" # "both" #"bbox_detection"
run_script = "Inference"  # "Inference" or "Training"
val_set = "Test"  # "Test" or "Val"
gpu_num = "0"

current_results_dir = "both_for_roots_2_notebook"
estimate_type = "withKeyPoints"
num_of_epochs = 2
have_gt = True
to_draw = False
load_weights = True
```

## Build Command

```python
def bool_arg(value: bool) -> str:
    """Convert a Python bool to the string expected by main.py."""
    return "true" if value else "false"


cmd = [
    sys.executable,
    str(MAIN_SCRIPT),
    "--storage-path", storage_path,
    "--dataset-name", dataset_name,
    "--network-type", network_type,
    "--run-script", run_script,
    "--val-set", val_set,
    "--gpu-num", gpu_num,
    "--current-results-dir", current_results_dir,
    "--estimate-type", estimate_type,
    "--num-of-epochs", str(num_of_epochs),
    "--have-gt", bool_arg(have_gt),
    "--to-draw", bool_arg(to_draw),
    "--load-weights", bool_arg(load_weights),
]

print(" ".join(f'"{part}"' if " " in part else part for part in cmd))
```

## Run

```python
result = subprocess.run(cmd, cwd=PROJECT_ROOT, check=False)
result.returncode
```
